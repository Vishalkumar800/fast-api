# 1. model_dump()
` Pydantic model ke data ko Python dictionary (dict) me convert karta hai.`

# Example:
```
  from pydantic import BaseModel

  class User(BaseModel):
      name: str
      age: int 

```

```
  user = User(name="Vishal", age=22)
 
  print(user.model_dump())
 
  Output:
  {'name': 'Vishal', 'age': 22}



  2. ** (Double Star)
  ** dictionary ko unpack karta hai.
  Dictionary ke key-value pairs ko keyword arguments me convert karta hai.

```

# Example:

  data = {
      "name": "Vishal",
      "age": 22
  }
#

```
# User(**data)
#
# Ye same hai:
#
# User(name="Vishal", age=22)


# 3. Simple Rule
#
# *  -> list / tuple ko unpack karta hai
# ** -> dictionary ko unpack karta hai

```

```
# 4. model_dump() + **
#
# Agar:
#
# user.model_dump()
#
# return karta hai:
#
# {'name': 'Vishal', 'age': 22}
#
# To:
#
# User(**user.model_dump())
#
# dictionary ko unpack karke:
#
# User(name="Vishal", age=22)
#
# bana deta hai.

```

# SHORT NOTE:
```
# model_dump() = Pydantic Model -> Dictionary
# **           = Dictionary -> Keyword Arguments  ```
