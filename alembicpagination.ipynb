Bilkul. Is code ko **pagination** ke example se samjho. `offset()` ko abhi side me rakho, pehle `page` aur `limit` samjho.

Tumhara code:

```python
def get_employee_service(page: int, limit: int, db: Session):

    skip = (page - 1) * limit

    employee = db.query(Employee).offset(skip).limit(limit).all()

    return employee
```

### Maan lo database me 100 employees hain

Aur tum chahte ho **ek page par 10 employees**.

```text
limit = 10
```

### Page 1

```text
page = 1
limit = 10
```

Ye line:

```python
skip = (page - 1) * limit
```

calculate karegi:

```text
skip = (1 - 1) × 10
     = 0
```

Matlab **0 employees skip karo**.

```python
.offset(0).limit(10)
```

Result:

```text
Employee 1
Employee 2
...
Employee 10
```

```text
page = 2
limit = 10
```

```text
skip = (2 - 1) × 10
     = 10
```

Matlab:

```python
.offset(10).limit(10)
```

```text
Employee 11
Employee 12
...
Employee 20
```

### Page 3

```text
page = 3
limit = 10
```

```text
skip = (3 - 1) × 10
     = 20
```

Matlab:

> Pehle 20 employees **skip** karo, phir 10 employees **lo**.

```python
.offset(20).limit(10)
```

Result:

```text
Employee 21
Employee 22
...
Employee 30
```

### Ab `offset()` ka actual meaning

```python
.offset(20)
```

ka matlab:

> **Database ke result me starting ke 20 records ko chhod do.**

Aur:

```python
.limit(10)
```

ka matlab:

> **Uske baad maximum 10 records do.**

So:

```python
db.query(Employee).offset(20).limit(10).all()
```

= **20 skip → next 10 lekar aao**

### Formula kyun hai?

```python
skip = (page - 1) * limit
```

Kyuki:

| Page | Skip | Records |
| ---- | ---: | ------- |
| 1    |    0 | 1–10    |
| 2    |   10 | 11–20   |
| 3    |   20 | 21–30   |
| 4    |   30 | 31–40   |

**Bas `offset` ko "kitne records pehle chhodne hain" samjho.** यही इसका पूरा concept है.
